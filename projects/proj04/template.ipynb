```python
# 1. Split players vs teams
players = worlds2025df[worlds2025df["position"] != "team"].copy()
teams   = worlds2025df[worlds2025df["position"] == "team"].copy()

# 2. Select all team-level columns you want to propagate
team_cols = [
    "gameid","side","teamname","teamid",
    "pick1","pick2","pick3","pick4","pick5",
    "ckpm","firstdragon","dragons","opp_dragons","elementaldrakes","opp_elementaldrakes",
    "infernals","mountains","clouds","oceans","chemtechs","hextechs",
    "dragons (type unknown)","elders","opp_elders",
    "firstherald","heralds","opp_heralds",
    "void_grubs","opp_void_grubs",
    "firstbaron","barons"
]

team_stats = teams[team_cols]

# 3. Merge back into player rows
players = players.merge(
    team_stats,
    on=["gameid","side","teamname","teamid"],
    suffixes=("", "_team")
)

summary = players[["gameid", "teamname", "wardsplaced", "wardskilled", "dragons_team"]].groupby(["gameid", "teamname"]).mean()
summary
```

```python
px.histogram(summary, x="wardsplaced", y="dragons_team")
```
