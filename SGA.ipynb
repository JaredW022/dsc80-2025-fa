```python
import pandas as pd

def get_unique_people(morning_line: pd.DataFrame, afternoon_line: pd.DataFrame) -> pd.DataFrame:
    # Combine both DataFrames
    combined = pd.concat([morning_line, afternoon_line], ignore_index=True)

    # Define identifying columns (adjust if needed)
    id_cols = ['First Name', 'Last Name', 'Email']

    # Drop duplicates based on identifying columns
    unique_people = combined.drop_duplicates(subset=id_cols, keep='first')

    return unique_people

original_list = list(get_unique_people(morning_line, afternoon_line)["First Name"].sort_values(ascending=True))
```
